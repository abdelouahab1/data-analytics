## Looping Through Arrays

```python
arrA = np.array([
    [11, 12, 13],
    [14, 15, 16]
])
```

```python
arrA
```

```python
# looping through the first dimension of arrA
for i in range(1):
    print(arrA[0] * 2)
```

```python
# looping through the second dimension of arrA
for i in range(1):
    print(arrA[1] * 2)
```

There is a 6-number difference between the first and second dimension after we duplicate both, So if we add 6 to the first array it's gonna equal the second array

```python
for x in arrA:
    loop_result = x*2
    print(loop_result[1])
```
