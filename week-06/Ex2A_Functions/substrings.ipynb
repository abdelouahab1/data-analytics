```python
def get_supplier_code(part_code):
    colon_position = part_code.find(':')
    supplier = part_code[:colon_position]
    return supplier
```

```python
get_supplier_code('DI:12345-M')
```

```python
example_1 = 'ACME:123-L'
example_2 = 'DI:12345-M'
example_3 = 'ACE:1-12'
```

```python
def get_product_num(part_code):
    colon_position = part_code.find(':')
    hyphen_position = part_code.find('-')
    product_num = part_code[colon_position + 1:hyphen_position]
    return product_num
```

```python
get_product_num(example_2)
```

```python
def get_size(part_code):
    hyphen_position = part_code.find('-')
    size = part_code[hyphen_position + 1:]
    return size
```

```python
get_size(example_3)
```

```python
print(f"For {example_1}, the supplier is {get_supplier_code(example_1)}. Part number {get_product_num(example_1)} is available in size {get_size(example_1)}.")
```
