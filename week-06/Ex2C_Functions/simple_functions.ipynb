# Lab 1 - Exercise 2.C: Custom Functions

In this lab, I will practice defining and calling custom functions
using parameters and return statements in Python.

```python
def favorite_things(name, movie, music):
    return f"My name is {name}.\nMy favorite movie is {movie}.\nMy favorite musician or band is {music}."

print(favorite_things("Abdul", "Interstellar", "Coldplay"))
```

```python
def why_im_here(purpose):
    return purpose

print(why_im_here("I joined the Year Up United Data Analyst program to develop my technical skills, gain professional experience, and build a career in data science."))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("New York City", "I enjoy its energy, diversity, and beautiful attractions"))
```

```python
print(favorite_place("Paris", "I love its architecture, culture, and history"))
```
